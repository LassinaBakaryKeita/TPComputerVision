```python
import cv2

# 1. Lire et afficher l'image d'origine
img = cv2.imread('image1.jpg') 
cv2.imshow('Image originale', img)
cv2.waitKey(0)

# 2. Conversion en niveaux de gris
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow('Image en niveaux de gris', gray)
cv2.waitKey(0)

# 3. Détection des contours avec l'algorithme de Canny
edges = cv2.Canny(img, 100, 200)
cv2.imshow('Contours détectés', edges)
cv2.waitKey(0)

# Fermer toutes les fenêtres à la fin
cv2.destroyAllWindows()
```

### Réponses - Exercice 1
**1. Que représente la matrice retournée par OpenCV pour l'image ?**
L'image est représentée en mémoire sous forme d'une matrice numérique (un tableau NumPy multi-dimensionnel) où chaque case correspond à un pixel. 
- Pour l'image originale en couleur (BGR), c'est une matrice 3D (Hauteur x Largeur x 3 canaux de couleurs). 
- Pour l'image en niveaux de gris, c'est une matrice 2D (Hauteur x Largeur) car un seul canal d'intensité lumineuse est suffisant.

**2. Pourquoi convertit-on souvent l'image en niveaux de gris avant traitement ?**
La conversion en niveaux de gris simplifie grandement l'image en réduisant les données de trois canaux à un seul. Cela permet de réduire les temps de calculs et la consommation mémoire des algorithmes (comme la détection de contours avec Canny), tout en conservant l'essentiel de l'information structurelle (formes, contrastes, arêtes).

```python
import cv2

# Lecture de la webcam (0) ou d'une vidéo
cap = cv2.VideoCapture(0) 

while True:
    ret, frame = cap.read()
    if not ret:
        break
        
    # Ajout du filtre pour convertir la vidéo en niveaux de gris en temps réel
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    cv2.imshow('Webcam - Niveaux de gris', gray_frame)
    
    # Ferme la vidéo quand l'utilisateur appuie sur 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### Réponse - Exercice 2
**Que se passe-t-il si vous utilisez une vidéo enregistrée à la place de la webcam ?**
Si l'on passe le chemin d'une vidéo (ex: `"video.mp4"`) à `cv2.VideoCapture()`, le programme va lire et traiter les frames de la vidéo le plus rapidement possible en fonction de la puissance du processeur, ce qui donne souvent un effet de lecture "en avance rapide" comparé au flux cadencé d'une webcam. Une fois la vidéo terminée, `cap.read()` renvoie `False`, ce qui coupe la boucle automatiquement.
